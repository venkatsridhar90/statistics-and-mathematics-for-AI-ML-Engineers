# Missing Values



# 1.isnull()

## Step 1 – Understand the Concept

`isnull()` is used to identify missing values in a Pandas DataFrame or Series.

A missing value means that some information is not available.

For example:

| Name | Age |
|---|---:|
| Arun | 22 |
| Priya | NaN |
| Kumar | 24 |

Here, Priya's Age is missing.

`isnull()` returns:
- `True` → value is missing
- `False` → value is not missing

It is one of the first methods used when checking the quality of a dataset.

## Step 2 – Demonstrate the Concept

### Simple Example

If a student's marks are missing, `isnull()` can identify that missing value.

### Real-World Example

A customer dataset may have missing phone numbers or email addresses.

### Business Example

A sales dataset may contain transactions where the sales amount is missing.

### AI/ML Use Case

Before training a machine learning model, we need to identify missing values in the features.

### Data Analytics Use Case

Analysts can use `isnull()` to understand which columns contain incomplete data.


## Step 3: Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar"],
    "Age": [22, None, 24],
    "Marks": [85, 90, None]
})

print(df.isnull())
print("\nMissing values in each column:")
print(df.isnull().sum())
```



# 2. notnull()


## Step 1 – Understand the Concept

`notnull()` is used to identify values that are available and not missing.

It is the opposite of `isnull()`.

It returns:

- `True` → value is available
- `False` → value is missing

This method is useful when we want to select only records that contain valid values.

## Step 2 – Demonstrate the Concept

### Simple Example

If a student's marks are available, `notnull()` returns `True`.

### Real-World Example

A customer database can be filtered to find customers whose email addresses are available.

### Business Example

A company can select sales records where the sales amount is available.

### AI/ML Use Case

`notnull()` can be used to select records with available feature values for analysis.

### Data Analytics Use Case

Analysts can filter datasets to work only with records containing required information.


## Step 3: Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar"],
    "Age": [22, None, 24],
    "Marks": [85, 90, None]
})

print(df.notnull())

print("\nRows where Age is available:")
print(df[df["Age"].notnull()])
```



# 3.dropna()

## Step 1 – Understand the Concept

`dropna()` is used to remove rows or columns containing missing values.

For example, if a row contains a missing value, `dropna()` can remove that row.

By default, `dropna()` removes rows containing at least one missing value.

It is useful when missing records are not required or when the number of missing values is small.

## Step 2 – Demonstrate the Concept

### Simple Example

If one student's marks are missing, we can remove that student's row using `dropna()`.

### Real-World Example

A survey dataset may contain a few incomplete responses that are not useful for analysis.

### Business Example

A company may remove transaction records where important required information is missing.

### AI/ML Use Case

Rows with missing values may sometimes be removed before model training when the number of missing records is very small.

### Data Analytics Use Case

Analysts can remove incomplete records when missing information is not important for the analysis.


## Step 3: Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar", "Anitha"],
    "Age": [22, None, 24, 25],
    "Marks": [85, 90, None, 88]
})

clean_df = df.dropna()

print(clean_df)
```



# 4.fillna()

## Step 1 – Understand the Concept

`fillna()` is used to replace missing values with another value.

Instead of removing a row containing missing data, we can fill the missing value.

For example:

Missing Age → 25

We can also use statistical values such as:
- Mean
- Median
- Mode

The choice depends on the type of data and the reason for the missing value.

## Step 2 – Demonstrate the Concept

### Simple Example

If one student's age is missing, we can fill it with a suitable value.

### Real-World Example

A customer dataset may have missing ages. The missing values can be filled using an appropriate statistical value.

### Business Example

A company may fill missing sales values using a carefully selected business rule or statistical method.

### AI/ML Use Case

Missing numerical features can sometimes be replaced using the mean or median before model training.

### Data Analytics Use Case

Analysts can fill missing values to avoid losing useful records during analysis.


## Step 3: Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar", "Anitha"],
    "Age": [22, None, 24, 25],
    "Marks": [85, 90, None, 88]
})

df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

print(df)
```



# 5. interpolate()

## Step 1 – Understand the Concept

`interpolate()` is used to estimate missing numerical values based on surrounding values.

Instead of simply filling a missing value with one fixed value, interpolation calculates a value based on the existing values around it.

For example:

10, 20, NaN, 40

The missing value can be estimated as:

30

Interpolation is especially useful when data follows an ordered sequence, such as time-series data.

## Step 2 – Demonstrate the Concept

### Simple Example

For the values:

10, 20, NaN, 40

interpolation can estimate the missing value as 30.

### Real-World Example

A temperature sensor may fail to record one reading. The missing temperature can be estimated from nearby readings.

### Business Example

A company may have a missing daily sales value in a time-series dataset and estimate it using nearby values.

### AI/ML Use Case

Interpolation can be used to fill missing values in time-dependent features before training a model.

### Data Analytics Use Case

Analysts can use interpolation to maintain continuous time-series data for trend analysis.


## Step 3: Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Day": [1, 2, 3, 4, 5],
    "Sales": [100, 120, None, 160, 180]
})

df["Sales"] = df["Sales"].interpolate()

print(df)
```



# 6. Missing Value Handling Strategies

## Step 1 – Understand the Concept

Missing value handling means deciding what to do when a dataset contains unavailable values.

There is no single method that is correct for every dataset.

Common strategies include:

1. Remove rows using `dropna()`
2. Remove columns when too many values are missing
3. Fill values using `fillna()`
4. Use mean, median, or mode
5. Use `interpolate()` for suitable numerical or time-series data
6. Keep missing values when the missingness itself has meaning

The correct strategy depends on:
- Amount of missing data
- Type of column
- Business meaning
- Data distribution
- Machine learning requirements

## Step 2 – Demonstrate the Concept

### Simple Example

If only one value is missing, we may fill it instead of removing the entire row.

### Real-World Example

If a sensor occasionally misses a reading, interpolation may be suitable.

### Business Example

If a customer phone number is missing, replacing it with the average value does not make sense. The missing value may need to remain missing or be handled using a business rule.

### AI/ML Use Case

Before training a model, missing values must be handled appropriately because many ML algorithms cannot work directly with missing values.

### Data Analytics Use Case

Analysts choose a missing-value strategy based on how much data is missing and how important the affected column is.

### Important Point

We should not automatically replace every missing value with zero.

The correct strategy depends on what the missing value represents.


## Step 3 – Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar", "Anitha", "Rahul"],
    "Age": [22, None, 24, 25, None],
    "Marks": [85, 90, None, 88, 76]
})

print("Original Data:")
print(df)

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Fill numerical missing values using median
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Marks"] = df["Marks"].fillna(df["Marks"].median())

print("\nAfter Handling Missing Values:")
print(df)
```
