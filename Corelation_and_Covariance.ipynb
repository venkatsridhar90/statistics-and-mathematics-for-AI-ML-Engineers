# Correlation



* Correlation measures the strength and direction of the relationship between two variables.

* In simple words, it tells us whether two things tend to change together.

### Real-time Example
* A company may compares

* Advertising spending
* Product sales

* If advertising spending increases and sales also increase, they may have a positive correlation.

### Formula

For Pearson correlation:






### Code Implementation

```python
import numpy as np

advertising = np.array([10, 20, 30, 40, 50])
sales = np.array([20, 30, 45, 50, 65])

correlation = np.corrcoef(advertising, sales)[0, 1]

print("Correlation:", correlation)
```

# Covariance

* Covariance tells us whether two variables tend to increase or decrease together.

* Positive covariance --> variables tend to move together.
* Negative covariance --> one tends to increase when the other decreases.
* Near-zero covariance --> no clear linear relationship.
### Real-time Example

* A company compares:

* Number of hours employees work
* Number of tasks completed

* If employees working more hours generally complete more tasks, covariance may be positive.

### Formula







```python
import numpy as np

hours = np.array([2, 4, 6, 8, 10])
tasks = np.array([5, 10, 15, 20, 25])

covariance = np.cov(hours, tasks, bias=True)[0, 1]

print("Covariance:", covariance)
```

# Positive Correlation


* Positive correlation means that when one variable increases, the other variable generally also increases.

### Real-time Example

* * Study hours and exam marks

* Generally:

* Study hours ↑
* Exam marks ↑

* Therefore, they may have a positive correlation.

### Formula

* Correlation coefficient:



* A value closer to +1 indicates a stronger positive linear relationship.

```python
import numpy as np

study_hours = np.array([1, 2, 3, 4, 5])
marks = np.array([40, 50, 60, 75, 85])

correlation = np.corrcoef(study_hours, marks)[0, 1]

print("Positive Correlation:", correlation)
```

# Negative Correlation

* Negative correlation means that when one variable increases, the other variable generally decreases.

### Real-time Example

* Product price and number of products sold

* Generally:

* Price ↑
* Number of products sold ↓

* This represents negative correlation.

### Formula



* A value closer to -1 indicates a stronger negative linear relationship.


```python
import numpy as np

price = np.array([10, 20, 30, 40, 50])
sales = np.array([100, 80, 60, 40, 20])

correlation = np.corrcoef(price, sales)[0, 1]

print("Negative Correlation:", correlation)
```

# No Correlation
* No correlation means there is no clear linear relationship between two variables.

* Changes in one variable do not consistently correspond to changes in the other.

### Real-time Example

* A company compares:

* Employee ID
* Monthly sales

* An employee ID does not normally determine sales performance, so there may be no meaningful correlation.

### Formula





```python
import numpy as np

employee_id = np.array([1, 2, 3, 4, 5])
sales = np.array([50, 20, 80, 30, 60])

correlation = np.corrcoef(employee_id, sales)[0, 1]

print("Correlation:", correlation)
```

# Pearson Correlation


* Pearson Correlation measures the strength and direction of a linear relationship between two numerical variables.

* The result ranges from:




### Real-time Example

* A company wants to know whether:

* Employee training hours
* Employee productivity

* have a linear relationship.

* Pearson correlation can be used.

### Formula





```python
import numpy as np
from scipy.stats import pearsonr

training_hours = np.array([2, 4, 6, 8, 10])
productivity = np.array([20, 30, 40, 55, 65])

correlation, p_value = pearsonr(training_hours, productivity)

print("Pearson Correlation:", correlation)
print("P-value:", p_value)
```

# Spearman Correlation
* Spearman Correlation measures the relationship between two variables based on their ranks.

* It is useful when the relationship is monotonic but not necessarily linear, or when the data is ordinal/ranked.

### Real-time Example

* A company collects:

* Customer satisfaction ranking
* Product quality ranking

### For example:

| Customer | Satisfaction Rank | Quality Rank |
| -------- | ----------------: | -----------: |
| A        |                 1 |            2 |
| B        |                 2 |            1 |
| C        |                 3 |            3 |
| D        |                 4 |            4 |
| E        |                 5 |            5 |


* Spearman correlation can measure the relationship between these rankings.

### Formula

* When there are no tied ranks:






$d_i$ = difference between the two ranks
$n$ = number of observations



```python
import numpy as np
from scipy.stats import spearmanr

satisfaction = np.array([1, 2, 3, 4, 5])
quality = np.array([2, 1, 3, 4, 5])

correlation, p_value = spearmanr(satisfaction, quality)

print("Spearman Correlation:", correlation)
print("P-value:", p_value)
```

# Correlation Matrix

* A Correlation Matrix shows the correlation between multiple variables at the same time.

* Each cell contains the correlation coefficient between two variables.

### Real-time Example

* A company has the following variables:

* Advertising
* Sales
* Profit
* Customer visits

* A correlation matrix can show the relationship between every pair of these variables.

### Formula

* The correlation coefficient between each pair of variables is calculated using a correlation method such as Pearson.


```python
import pandas as pd

data = {
    "Advertising": [10, 20, 30, 40, 50],
    "Sales": [20, 30, 45, 50, 65],
    "Profit": [5, 8, 12, 15, 20],
    "Customers": [100, 150, 200, 250, 300]
}

df = pd.DataFrame(data)

correlation_matrix = df.corr()

print(correlation_matrix)
```

# Heatmap

* A Heatmap is a visual representation of numerical values using different shades or colors.

* For correlation analysis, a heatmap makes it easy to identify strong positive, strong negative, and weak relationships.

### Real-time Example

* A company has many features such as:

* Age
* Income
* Spending
* Purchase frequency
* Website visits

* A correlation heatmap can quickly show which features are strongly related.


```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Advertising": [10, 20, 30, 40, 50],
    "Sales": [20, 30, 45, 50, 65],
    "Profit": [5, 8, 12, 15, 20],
    "Customers": [100, 150, 200, 250, 300]
}

df = pd.DataFrame(data)

correlation_matrix = df.corr()

plt.imshow(correlation_matrix)
plt.colorbar()
plt.xticks(range(len(df.columns)), df.columns, rotation=45)
plt.yticks(range(len(df.columns)), df.columns)
plt.title("Correlation Heatmap")
plt.show()
```
