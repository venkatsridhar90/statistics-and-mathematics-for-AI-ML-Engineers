# Hypothesis Testing 

# 1.Null Hypothesis (H₀)
*  The null hypothesis says there is no significant change, difference, or effect in the situation being tested.
*   It is the starting assumption that we try to find evidence against.

### real time example
* A food-delivery company claims its average delivery time is 30 minutes. H₀ says the true average delivery time is 30 minutes.

### formula 
$$ H₀: μ = 30 $$

```python
claim = 12

print("Null Hypothesis: Average battery life =", claim, "hours")
```

# 2. Alternative Hypothesis (H₁ / Hₐ)
* The alternative hypothesis says that a real change, difference, or effect exists.
*  It is what we support when the sample provides enough evidence against H₀.

### Real-time example
* If the company wants to check whether the average delivery time is different from 30 minutes, H₁ says it is not 30 minutes.

### Formula:H₁: 
$$ μ ≠ 30 $$

```python
print("Alternative Hypothesis: Average battery life is not 12 hours")
```

# 3. p-value
* The p-value measures how unusual the observed result would be if the null hypothesis were true. A small p-value gives stronger evidence against H₀.

### Real-time example:
* A website changes its checkout page and gets a p-value of 0.02 for conversion improvement. At α = 0.05, the result is statistically significant.

### Formula:
$$Reject H₀ if p-value ≤ α$$ 


```python
p_val = 0.03

if p_val < 0.05:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")
```

## 4.Confidence Interval
* A confidence interval gives a range of reasonable values for a population parameter based on sample data.
* It shows the uncertainty around the estimated value.

### Real-time example:
* A survey estimates that 62% of customers are satisfied, with a 95% confidence interval of 58%–66%.

### Formula:
$$ CI = Estimate ± Critical Value × Standard Error$$

```python
import numpy as np

sample = [9.5, 12.2, 9.8, 12.1, 9.7, 12.3, 9.9, 12.0, 9.6, 12.1]

print("Mean:", np.mean(sample))
```

# 5. Significance Level (α)
 * The significance level is the maximum probability of making a Type I error that we are willing to accept.
 * It is chosen before performing the hypothesis test.

### Real-time example:
* A pharmaceutical company uses α = 0.05 when testing whether a new treatment has an effect.

### Formula:
$$α = P(Type I Error)$$

```python
alpha = 0.06

print("Significance Level:", alpha)
```

## 6.Type I Error
* A Type I error happens when we reject the null hypothesis even though it is actually true.
* In simple terms, it is a false alarm.

### Real-time example:
* A quality-control system says a good batch of products is defective and rejects it.

### Formula:
$$P(Type I Error) = α$$

```python
print("Type I Error: Rejecting a true Null Hypothesis")
```

## 7.Type II Error
* A Type II error happens when we fail to reject the null hypothesis even though it is actually false. In simple terms, we miss a real effect.

### Real-time example:
* A quality-control system accepts a defective batch of products as good.

### Formula:
$$P(Type II Error) = β$$

```python
print("Type II Error: Failing to reject a false Null Hypothesis")
```

## 8. One-tailed Test
* A one-tailed test checks for a difference in only one specified direction, such as greater than or less than a target.
* It is used when only one direction matters.

### Real-time example:
* A company wants to know whether a new training program increases productivity above 100 units per day.

### Formula:
$$H₁: μ > μ₀ (right-tailed)$$ or $$H₁: μ < μ₀ (left-tailed)$$

```python
observed_mean = 12.2

if observed_mean > 12:
    print("Supports Right-Tailed Test")
```

## 9. Two-tailed Test
* A two-tailed test checks whether a value is different from a target in either direction.
* It is used when both an increase and a decrease are important.

### Real-time example:
* A manufacturer wants to know whether the average battery life is different from the advertised 10 hours, whether higher or lower.

### Formula:
$$ H₁: μ ≠ μ₀$$

```python
observed_mean = 9.7

if observed_mean != 12:
    print("Supports Two-Tailed Test")
```
