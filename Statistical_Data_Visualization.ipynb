# Statistical Data Visualization

```python
### Sample dataset
import pandas as pd

data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug'],
    'Sales': [120, 135, 128, 150, 165, 155, 180, 195],
    'Ad_Spend': [20, 25, 22, 30, 35, 32, 40, 45],
    'Customers': [100, 110, 105, 125, 135, 130, 150, 160],
    'Profit': [25, 30, 28, 35, 40, 38, 45, 50],
    'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West']
}

df = pd.DataFrame(data)

print(df)
```

# Histogram

#### Purpose: 
* Shows the distribution of numerical data by grouping values into ranges (bins).
#### When to use:
* Use it to understand the shape, spread, center, and possible outliers of numerical data.
#### Advantages:
* Easy to see distribution, skewness, and concentration of values.
#### Limitations:
* Results depend on the chosen bin size; it is not suitable for categorical data.



```python

import matplotlib.pyplot as plt

plt.hist(data)
plt.show()
```

# Bar Chart
#### Purpose: 
* Compares values across different categories.
#### When to use: 
* Use it for comparing sales, employees, products, regions, or other categorical values.
#### Advantages: 
* Simple and easy to compare categories.
#### Limitations: 
* Too many categories can make the chart difficult to read.
### Example

```python
plt.bar(df['Region'], df['Sales'])
plt.show()
```

# Line Chart
#### Purpose:
* Shows how a value changes over time or across an ordered sequence.
#### When to use:
* Use it for monthly sales, yearly revenue, daily website traffic, or stock prices.
#### Advantages:
* Clearly shows trends, increases, decreases, and patterns over time.
#### Limitations:
* Not ideal for comparing many unrelated categories.
### Example:

```python
plt.plot(df['Month'], df['Sales'])
plt.show()
```

# Scatter Plot
#### Purpose: 
* Shows the relationship between two numerical variables.
#### When to use: 
* Use it to check whether two variables have a positive, negative, or weak relationship.
#### Advantages: 
* Helps identify correlation, patterns, clusters, and outliers.
#### Limitations: 
* It does not prove that one variable causes another.

```python
plt.scatter(df['Ad_Spend'], df['Sales'])
plt.xlabel('Ad Spend')
plt.ylabel('Sales')
plt.show()
```

5. Pie Chart
* Shows how different categories contribute to a total.
#### When to use:
* Use it when showing percentages or market/revenue shares with a small number of categories.
#### Advantages:
*  Easy to understand part-to-whole relationships.
#### Limitations:
*  Difficult to compare similar-sized slices and becomes confusing with many categories.
### Example:

```python
plt.pie(df['Sales'], labels=df['Month'],autopct="%1.1f%%")

plt.show()
```

# Density Plot
*  Shows the smooth distribution of numerical data.
#### When to use:
*  Use it to understand the shape of data and compare distributions between groups.
#### Advantages:
*  Provides a smooth view of the distribution and is useful for comparing groups.
#### Limitations:
*  The appearance depends on the smoothing method, and it can be harder for beginners to understand.

### Example:

```python
import seaborn as sns

sns.kdeplot(df['Sales'])
plt.show()
```

# Box Plot
* Shows the median, spread, quartiles, and possible outliers in numerical data.
### When to use:
*  Use it to compare distributions between different groups.
### Advantages:
* Quickly identifies outliers and compares the spread of multiple groups.
### Limitations:
*  It does not show the detailed shape of the distribution.

```python
sns.boxplot(x=df['Sales'])
plt.show()
```

# Violin Plot
* Combines a box plot with a density plot to show both summary statistics and distribution shape.
#### When to use: 
* Use it when you want to compare the distribution and spread of numerical data across groups.
#### Advantages: 
* Shows more information about the distribution than a basic box plot.
#### Limitations: 
* Can be difficult to interpret for beginners and depends on density smoothing.

```python
sns.violinplot(x=df['Region'], y=df['Sales'])
plt.show()
```

# Pair Plot
* Shows relationships and distributions between multiple numerical variables at the same time.
#### When to use:
* Use it during exploratory data analysis to quickly understand relationships among several features.
#### Advantages:
* Gives a quick overall view of relationships, distributions, and possible patterns.
#### Limitations:
* Becomes crowded and difficult to read when there are many features.

```python
sns.pairplot(df[['Sales', 'Customers', 'Profit']])
plt.show()
```

# Heatmap
* Uses colors to represent values in a matrix, making patterns easier to identify.
### When to use:
* Commonly used for correlation matrices, missing values, and categorical/count tables.
### Advantages:
* Quickly highlights strong, weak, high, and low values.
### Limitations:
* Too many variables can make it crowded, and color differences may be difficult to interpret.

```python
sns.heatmap(
    df[['Sales', 'Customers', 'Profit']].corr(),
    annot=True
)
plt.show()
```
