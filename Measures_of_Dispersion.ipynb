## Measures of dispersion
* Measures of dispersion are statistical methods used to describe how much the data values are spread out or scattered from the center
* There are five most commonly used measures of dispersion.
*  These are range, variance, standard deviation, mean deviation, and quartile deviation.
## why it is important
* understand data variability
* compare datasets
* identify consistency
* supports decision making
# Range:
* The range is the difference between the largest value and the smallest value in a dataset.
* It is one of the simplest ways to understand how spread out the data is.
## formula 
* Range = Maximun - Minimum

### real time example
* Suppose if a student marks are
* 48,49,50,52,54
* here range = 54 - 48
* range = 6
### example:


```python
import numpy as np

marks = np.array([40, 50, 60, 70, 80])

data_range = np.max(marks) - np.min(marks)

print("Range:", data_range)
```

# Variance
* Variance measures how far the values in a dataset are spread from the mean.
* It first calculates the difference between each value and the mean, squares those differences, and then calculates their average.
* A small variance means the values are close to the mean.
* A large variance means the values are more spread out.
## formula



where:

$x_i$ = individual value
μ = population mean
$N$ = number of observations

### Numerical example
2, 4, 6
mean = 4

$$ \frac{2+4+6}{3}=4 $$
Squared differences:
(2 - 4)² = 4 

(4 - 4)² = 0

(6 - 4)² = 4

Variance:
$$ \frac{4+0+4}{3} = \frac{8}{3} $$
Variance = 2.67

### real time example
* Suppose two stores have the following daily sales.
* Store A:
* ₹9,000, ₹10,000, ₹11,000
* Store B:
* ₹2,000, ₹10,000, ₹18,000
* Both have a mean of ₹10,000, but Store B has much greater variation.
* Therefore, Store B has a higher variance.

```python
import numpy as np

data = np.array([2, 4, 6])

variance = np.var(data)

print("Variance:", variance)
```

# Standard Deviation
* Standard deviation tells us how much the values typically vary from the mean.
* It is closely related to variance.
## formula


for example the data is 2,4,6

variance = 2.67

therefore,standard deviation = square root of variance

standard deviation = $\sqrt{2.67}$.

standard deviation = 1.63

### real time example
* Suppose two machines produce parts.
* Machine A produces sizes close to the required size.
* Machine B produces parts with much more variation.
* The machine with the smaller standard deviation produces more consistent results.
### code implementation


```python
import numpy as np

data = np.array([2, 4, 6])

standard_deviation = np.std(data)

print("Standard Deviation:", standard_deviation)
```

# Quartiles
* Quartiles divide ordered data into four approximately equal parts.
* There are three important quartiles:
* Q1 → 25th percentile
* Q2 → 50th percentile / Median
* Q3 → 75th percentile
### formula


for example

10, 20, 30, 40, 50, 60, 70

Using percentile calculation:

Q1 = 25th percentile
Q2 = 50th percentile
Q3 = 75th percentile

Q1 = 25 → approximately 25% of the data lies below this point.

Q2 = 40 → median.

Q3 = 55 → approximately 75% of the data lies below this point.

```python
import numpy as np

data = np.array([10, 20, 30, 40, 50, 60, 70])

q1 = np.percentile(data, 25)
q2 = np.percentile(data, 50)
q3 = np.percentile(data, 75)

print("Q1:", q1)
print("Q2:", q2)
print("Q3:", q3)
```

# Percentiles
* A percentile tells us the position of a value relative to the rest of the dataset.
* For example, the 90th percentile is a value below which approximately 90% of the observations fall.
### formula 
* $P_k$ = value below which approximately k% of observations fall

### example

40, 50, 60, 70, 80, 90, 100

The 50th percentile represents the middle of the distribution.

$$ P_{50} = 70 $$
### code implementation

```python
import numpy as np

scores = np.array([40, 50, 60, 70, 80, 90, 100])

p1 = np.percentile(scores, 25)
p2 = np.percentile(scores, 50)
p3 = np.percentile(scores, 90)

print("25th Percentile:", p1)
print("50th Percentile:", p2)
print("90th Percentile:", p3)
```

# IQR(Interquartile Range)

* it measures the spread of the middle 50% of the data.
* It ignores the lowest 25% and highest 25% of observations.
### Formula 
IQR = $Q_3$ - $Q_1$

where $Q_3$ = 75th percentile
$Q_1$ = 25th percentile

example:

Q1 = 25
Q3 = 55

Then:

$$ IQR = 55-25 $$
IQR=30

### example

```python
import numpy as np

data = np.array([10, 20, 30, 40, 50, 60, 70])

q1 = np.percentile(data, 25)
q3 = np.percentile(data, 75)

iqr = q3 - q1

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
```

# Coefficient of variation
* The Coefficient of Variation (CV) measures the amount of variation relative to the mean.
* Unlike standard deviation, CV is unit-free, so it can be useful for comparing the variability of datasets with different scales.
### Formula


for example

Mean = 50, 
Standard Deviation = 5

Then:

$$ CV = \frac{5}{50}\times100 $$
CV=10%

```python
import numpy as np

data = np.array([40, 45, 50, 55, 60])

mean = np.mean(data)
standard_deviation = np.std(data)

cv = (standard_deviation / mean) * 100

print("Mean:", mean)
print("Standard Deviation:", standard_deviation)
print("Coefficient of Variation:", cv, "%")
```
