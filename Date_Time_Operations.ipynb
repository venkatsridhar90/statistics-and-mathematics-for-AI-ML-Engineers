# Date & Time Operations



# 1. Datetime


## Step 1 – Understand the Concept

* Datetime means a value that contains both a date and a time.

* For example:

* 2026-10-08 09:30:00

Here:
- 2026 → Year
- 10 → Month
- 08 → Day
- 09 → Hour
- 30 → Minute
- 00 → Second

* In Pandas, `pd.to_datetime()` is used to convert date and time values into a proper datetime format.

* Datetime is important because many real-world datasets contain information about when an event happened.

* For example, sales, customer orders, website visits, payments, and sensor readings all have date and time information.

## Step 2 – Demonstrate the Concept

### Simple Example

A value like `2026-10-08 09:30:00` tells us the exact date and time.

### Real-World Example

A hospital can store the exact date and time when a patient visits the hospital.

### Business Example

An online shopping company can store the date and time when a customer places an order.

### AI/ML Use Case

Datetime values can be used to create features such as month, day, hour, or weekday for predicting sales.

### Data Analytics Use Case

An analyst can use datetime values to study daily, weekly, or monthly business trends.


## Step 3: Implement the Concept

```python
import pandas as pd

date_time = pd.to_datetime("2026-10-08 09:30:00")

print(date_time)
print(type(date_time))
```



# 2.Timestamp


## Step 1 – Understand the Concept

A Timestamp represents one specific point in time.

For example:

2026-10-08 10:15:30

This tells us exactly when an event happened.

Pandas provides `pd.Timestamp()` to create a Timestamp.

A Timestamp is useful when we need to record or compare the exact time of events.

For example, transactions, logins, payments, sensor readings, and system activities can be stored using timestamps.

## Step 2 – Demonstrate the Concept

### Simple Example

`2026-10-08 10:15:30` represents one exact point in time.

### Real-World Example

A mobile application can record the exact time when a user logs in.

### Business Example

A bank can store the exact timestamp when a customer makes a transaction.

### AI/ML Use Case

Timestamps can help an ML model understand when customer activities happen.

### Data Analytics Use Case

Analysts can calculate how many transactions happened during a particular hour.


## Step 3: Implement the Concept

```python
import pandas as pd

timestamp = pd.Timestamp("2026-10-08 10:15:30")

print(timestamp)
print("Year:", timestamp.year)
print("Month:", timestamp.month)
print("Day:", timestamp.day)
```



# 3.Date Parsing


## Step 1 – Understand the Concept


Date parsing means converting a date stored as text into a proper datetime format.

Dates can appear in different formats.

For example:

- 2026-10-08
- 08-10-2026
- 2026/10/08

Computers may treat these values as text if they are not converted properly.

Pandas provides `pd.to_datetime()` to parse date values.

Date parsing is important because it makes date values easier to compare, calculate, and analyze.

## Step 2 – Demonstrate the Concept

### Simple Example

The text `"2026-10-08"` can be converted into a datetime value.

### Real-World Example

A company may receive customer data from different systems with different date formats.

### Business Example

Sales data from different branches may contain dates in different formats. Date parsing helps convert them into a consistent datetime format.

### AI/ML Use Case

Date parsing is an important preprocessing step before creating date-based ML features.

### Data Analytics Use Case

Parsed dates can be used to group sales by month, compare dates, and calculate time differences.

## Step 3: Implement the Concept

```python
import pandas as pd

dates = pd.Series([
    "2026-01-15",
    "2026-02-20",
    "2026-03-25"
])

parsed_dates = pd.to_datetime(dates)

print(parsed_dates)
```



# 4.Timedelta

## Step 1 – Understand the Concept

Timedelta represents the difference or duration between two dates or times.

For example:

- 5 days
- 3 hours
- 30 minutes

If an order is placed on October 1 and delivered on October 5, the delivery duration is 4 days.

Pandas uses Timedelta to represent this type of duration.

Timedelta is useful when we need to calculate how much time has passed between two events.

## Step 2 – Demonstrate the Concept

### Simple Example

If an order takes 3 days to arrive, the delivery time is a Timedelta of 3 days.

### Real-World Example

A food delivery application can calculate how long an order takes from placement to delivery.

### Business Example

An e-commerce company can calculate the average delivery time for its customers.

### AI/ML Use Case

Delivery duration can be used as a feature to predict late deliveries.

### Data Analytics Use Case

An analyst can calculate average processing time, delivery time, or customer waiting time.


## Step 3: Implement the Concept

```python
import pandas as pd

start_date = pd.Timestamp("2026-10-01")
end_date = pd.Timestamp("2026-10-08")

duration = end_date - start_date

print("Duration:", duration)
print("Days:", duration.days)
```



# 5.Time Zones


## Step 1 – Understand the Concept

A time zone represents the local time used in a particular geographical region.

For example:

India → IST
United Kingdom → GMT/BST
United States → Multiple time zones

The same moment can have different local times in different countries.

Pandas allows us to work with timezone-aware datetime values.

Time zones are important when data comes from different countries or locations.

## Step 2 – Demonstrate the Concept

### Simple Example

The same moment can have different local times in India and the United States.

### Real-World Example

A video meeting scheduled in India may occur at a different local time for a person in another country.

### Business Example

A global e-commerce company may receive orders from customers in different countries.

### AI/ML Use Case

Correct timezone information helps ML systems understand when customer activities actually occurred.

### Data Analytics Use Case

Analysts can convert timestamps into a common timezone before comparing events from different regions.


## Step 3: Implement the Concept

```python
import pandas as pd

india_time = pd.Timestamp(
    "2026-10-08 09:00:00",
    tz="Asia/Kolkata"
)

print("India Time:", india_time)

utc_time = india_time.tz_convert("UTC")

print("UTC Time:", utc_time)
```



# 6.Extracting Date Features

## Step 1 – Understand the Concept

Date feature extraction means taking useful information from a datetime value.

For example, from:

2026-10-08 09:30:00

we can extract:

- Year → 2026
- Month → 10
- Day → 8
- Hour → 9
- Quarter → 4

Pandas provides the `.dt` accessor to extract these features from datetime columns.

Date feature extraction is especially useful in data analysis and machine learning because individual date components can reveal useful patterns.

## Step 2 – Demonstrate the Concept

### Simple Example

From `2026-10-08 09:30:00`, we can extract the year, month, day, hour, and quarter.

### Real-World Example

A shopping application can identify which day or hour receives the most orders.

### Business Example

A company can identify which months have the highest sales and plan inventory accordingly.

### AI/ML Use Case

Year, month, day, hour, and weekday can be used as features for sales prediction models.

### Data Analytics Use Case

Analysts can compare sales by month, weekday, quarter, or hour to identify business trends.


## Step 3: Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Order_Date": pd.to_datetime([
        "2026-01-15 10:30:00",
        "2026-05-20 14:45:00",
        "2026-10-08 09:15:00"
    ]),
    "Sales": [5000, 7500, 9000]
})

df["Year"] = df["Order_Date"].dt.year
df["Month"] = df["Order_Date"].dt.month
df["Day"] = df["Order_Date"].dt.day
df["Day_Name"] = df["Order_Date"].dt.day_name()
df["Hour"] = df["Order_Date"].dt.hour
df["Quarter"] = df["Order_Date"].dt.quarter

print(df)
```
