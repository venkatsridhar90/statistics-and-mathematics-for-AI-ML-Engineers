## 1. Normal Distribution

### Definition

* A Normal Distribution is a continuous probability distribution with a characteristic bell-shaped curve.
* where most of the data values are concentrated around the center (mean), and fewer values occur as we move away from the center.


In a normal distribution:

- Mean = Median = Mode
- Most observations are near the mean.
- Very few observations occur far from the mean.
- The left and right sides of the distribution have the same shape.

### Numerical Example

* heights of a students are

* 151,153,155,165,167,159,170,160,173

* Most values are concentrated around the center, while very small and very large values occur less frequently.

* This type of pattern can be represented approximately by a normal distribution.

### Example

* Employee performance scores in a large organization may approximately follow a normal distribution.
* Most employees may have average performance scores, while only a smaller number have extremely high or extremely low scores.

### Formula



```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(70, 10, 1000)

plt.hist(data, bins=30)
plt.xlabel("Scores")
plt.ylabel("Frequency")
plt.title("Normal Distribution")
plt.show()
```

## 2. Uniform Distribution

### Definition

* A uniform distribution is a distribution where values within a specified range have equal probability density.
* In a continuous uniform distribution, every interval of the same length has the same probability

###  Example

* Suppose a random number generator produces numbers between 1 and 10.

* Each number has approximately the same probability of being selected.

### Formula


### real time example
* Suppose a system randomly assigns customers a discount value between 0% and 10%.

* If the values are generated uniformly, no part of the range is intentionally favored.

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.uniform(0, 10, 1000)

plt.hist(data, bins=20)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Uniform Distribution")
plt.show()
```

## 3. Gaussian Distribution

### Definition

* A Gaussian Distribution is another name for the Normal Distribution.

* It is a continuous probability distribution characterized by its bell-shaped and symmetric curve.

* The shape of the Gaussian distribution is determined mainly by:

- Mean – determines the center of the distribution.
- Standard deviation – determines how spread out the data is.

### Formula



### Numerical Example
* Suppose the delivery time of a product is approximately Gaussian:

* Mean delivery time = 3 days
* Standard deviation = 0.5 days

* Most deliveries will be close to 3 days.


```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(100, 15, 1000)

plt.hist(data, bins=30)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Gaussian Distribution")
plt.show()
```

## 4. Left Skew

### Definition

* A Left-Skewed Distribution is a distribution where the long tail extends toward the left side.

* Most of the observations are concentrated toward the higher values, while a smaller number of low values create the long left tail.

* In a typical left-skewed distribution:
* Mean < Median < Mode

### Numerical Example

* Suppose a company has employee performance scores where most employees score highly, but a few employees have very low scores.
* The low scores create a left tail.



```python
left_skew = np.random.beta(5, 2, 100) * 100

plt.hist(left_skew, bins=10)
plt.title("Left Skewed Distribution")
plt.show()
```

## 5. Right Skew

### Definition

* A Right-Skewed Distribution is a distribution where the long tail extends toward the right side.

* Most observations are concentrated toward the lower values, while a smaller number of high values create the long right tail.

* In a typical right-skewed distribution:

Mean > Median > Mode

### Numerical Example
Consider:

10, 12, 14, 15, 16, 18, 80

Most values are relatively small, but 80 creates a long right tail.

### example 
Many people may have moderate incomes, while a small number of people have extremely high incomes.

Those high incomes create a right tail.

```python
right_skew = np.random.beta(2, 5, 100) * 100

plt.hist(right_skew, bins=10, edgecolor="black")
plt.title("Right Skewed")
plt.show()
```

## 7. Kurtosis
### Definition

* Kurtosis is a statistical measure that describes the shape of the tails of a distribution
* how much extreme values may occur compared with a normal distribution.
* It helps us understand whether a distribution has relatively heavy or light tails.

There are three common descriptions:

- Mesokurtic – similar tail behavior to a normal distribution.
- Leptokurtic – heavier tails and more extreme values.
- Platykurtic – lighter tails and fewer extreme values.

### formula
* 

```python
import numpy as np
from scipy.stats import kurtosis

data = np.array([10, 12, 13, 14, 15, 16, 18, 20, 50])

print("Mean:", np.mean(data))
print("Excess Kurtosis:", kurtosis(data))
```

## 8. Empirical Rule (68–95–99.7)

### Definition

* The Empirical Rule describes how the data is distributed around the mean in an approximately normal distribution.

* For a normal distribution:

- Approximately 68% of data falls within 1 standard deviation of the mean.
- Approximately 95% falls within 2 standard deviations.
- Approximately 99.7% falls within 3 standard deviations.

This is also called the 68–95–99.7 rule.

### Formula
For normal distribution


contains approximately 68% of observations.


contains approximately 95%.


1 Standard Deviation
$$ 100-10 = 90 $$ $$ 100+10 = 110 $$

Approximately 68% of observations lie between:

90 and 110
2 Standard Deviations
$$ 100-20=80 $$ $$ 100+20=120 $$

Approximately 95% lie between:

80 and 120
3 Standard Deviations
$$ 100-30=70 $$ $$ 100+30=130 $$

Approximately 99.7% lie between:

70 and 130



```python
import numpy as np

mean = 50
standard_deviation = 12

range_1 = (mean - standard_deviation, mean + standard_deviation)
range_2 = (mean - 2 * standard_deviation, mean + 2 * standard_deviation)
range_3 = (mean - 3 * standard_deviation, mean + 3 * standard_deviation)

print("Within 1 Standard Deviation:", range_1)
print("Within 2 Standard Deviations:", range_2)
print("Within 3 Standard Deviations:", range_3)
```
