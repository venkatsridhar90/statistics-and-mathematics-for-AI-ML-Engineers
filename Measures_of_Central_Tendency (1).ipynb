# Measures of Central Tendency

### what is Measures of Central Tendency?
* Measures of central tendency are statistical values used to describe the center or typical value of a set of data.
* it help us find a single value that represents the whole dataset.

### Why are they important?
* help us to summarize a large amount of data using one representative value.
* Useful for decision-making
* Helps in comparison
* Used in Business Analytics, Finance, Healthcare, and AI/ML.

 ## Common measures are
* Mean
* Median
* Mode
* Weighted Mean
* Geometric Mean
* Harmonic Mean

# Mean
* The mean is the average of all values in a dataset.
* We calculate it by adding all values and dividing the total by the number of values.
* The mean is useful when we want one value that represents the overall level of the data.

Formula:

Mean is frequently used in data analysis, preprocessing, normalization, and handling missing numerical values.

* Σx = Sum of all observations
* n = Number of observations

Example:50,60,40,70,80

Sum = 50+60+40+70+80 = 300

Mean =300 / 5 = 60

Mean = 60

```python
import numpy as np

marks = [50,60,40,70,80]

print(np.mean(marks))
```

# Median
* The median is the middle value after arranging the data in ascending or descending order.
* The median is especially useful when the dataset contains outliers
*  because extreme values have less influence on it than they do on the mean.
*  for odd numbers middle value is the median
*  for even numbers the average of two middle values is median

### Formulas 
* for odd numbers

* for even observations:

* Odd Example:
11,23,45,67,89

Middle value = 45

Median = 45

* Even example:
10, 20, 30, 40,50,60

Median:
(30 +40)/2 = 35

```python
import numpy as np

numbers = [11,23,45,67,89]

print(np.median(numbers))

number = [ 10,20,30,40,50,60]

print(np.median(number))
```

# Mode
 * The mode is the value that occurs most frequently in a dataset.

* A dataset can have:

* One mode --> Unimodal
* Two modes --> Bimodal
* More than two modes --> Multimodal
* No repeated value --> No mode
#### Formula
There is no mathematical calculation formula like mean.
*  Mode = most frequently occured value

Numerical Example 
Dataset:
2, 4, 5, 5, 6, 7, 5

Mode = 5

```python
import pandas as pd

numbers = pd.Series([2,4,5,5,6,7,5])

print(numbers.mode())
```

# Weighted Mean
* A weighted mean is an average where different values have different levels of importance.
* In a normal mean, every value has equal importance.
* In a weighted mean, some values contribute more than others.
#### Formula:

Here,
$x_i$ = value
$w_i$ = weight

Example:

Calculation :

(80×0.20)+(70×0.30)+(90×0.50)
=16+21+45 
=82

```python
import numpy as np

marks = [80,70,90]
weights = [20,30,50]

print(np.average(marks, weights=weights))
```

# Harmonic Mean
* The harmonic mean is an average that is particularly useful for rates and ratios.
* It gives more importance to smaller values than the arithmetic mean.
* It is commonly used for speeds,rates,ratios and precison and recall
### Formula

#### Example :
speeds = 60km/ph,40km/ph

Harmonic Mean = 2/(1/60 + 1/40)
Harmonic mean = 48 km/ph

```python
from scipy.stats import hmean

numbers = [40,60]

print(hmean(numbers))
```
