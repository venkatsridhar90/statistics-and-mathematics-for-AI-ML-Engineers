# Probability

### What is probability?

* Probability is the measure of how likely an event is to happen.

* It ranges from 0 to 1, where 0 means impossible and 1 means certain.

### Formula:

P(E)= Number of Favorable Outcomes / Total Number of Outcomes

### Example:

Tossing a fair coin

Possible outcomes: {Head, Tail}

P(Head) = 1/2 = 0.5

```python
import random

coin = random.choice(["Head", "Tail"])

print("Coin toss:", coin)
print("P(Head) =", 1/2)
```

### Events

* An event is a specific outcome or set of outcomes in a probability experiment.

### Example:

rolling a dice

Event A:

Getting a number divisible by 3 : {3, 6}

Probability : 2/6 = 0.333

```python
event = [x for x in range(1, 7) if x % 2 == 0]
print(event)
print(len(event) / 6)
```

# Sample Space

* A sample space is the complete set of all possible outcomes of a probability experiment.
* It is usually represented by the symbol S.

### Example:

Tossing two coins

S = {HH, HT, TH, TT}

```python
sample_space = ["HH", "HT", "TH", "TT"]

print(sample_space)
print("Total outcomes:", len(sample_space))
```

# Independent Events

* Independent events are events where one event does not affect the occurrence of another.
* The outcome of one event has no influence on the probability of the other event.

### Example:

Toss a coin two times.

The first toss does not change the result of the second toss.

P(Head and Head) = 1/2 × 1/2 = 1/4

```python
import random

toss1 = random.choice(["H", "T"])
toss2 = random.choice(["H", "T"])

print(toss1, toss2)
print("P(Head and Head) =", 1/2 * 1/2)
```

# Dependent Events

* Dependent events are events where the occurrence of one event affects the occurrence of another event.
* The outcome of one event changes the probability of the next event.

### Example:

A bag has 4 red balls and 3 blue balls.

Two balls are drawn one after another without putting the first one back.

P(both red) = 4/7 × 3/6 = 2/7

```python
balls = ["R"] * 4 + ["B"] * 3

p_first = balls.count("R") / len(balls)

balls.remove("R")

p_second = balls.count("R") / len(balls)

print(p_first, p_second)
print("P(both red) =", p_first * p_second)
```

# Conditional Probability

* Conditional probability is the probability of an event occurring when another event is already known to have occurred.
* It is used to find the probability under a specific condition or situation.

### Formula:

P(A|B) = P(A ∩ B) / P(B)

### Example:

A dice is rolled and the number is known to be odd.

What is the probability that it is a prime number?

Odd numbers : {1, 3, 5}

Prime numbers among them : {3, 5}

Probability : 2/3

```python
event = [1, 3, 5]

prime = [x for x in event if x in [2, 3, 5]]

print(len(prime) / len(event))
```

# Bayes' Theorem

* Bayes' Theorem is used to calculate the probability of an event based on previously known information.
* It helps update the probability of an event when new evidence or information is available

### Formula:

P(A|B) = P(B|A) × P(A) / P(B)

### Example:

Suppose 20% of emails are spam, and the word "free" appears in 60% of spam emails.
Overall, the word "free" appears in 16% of all emails.

P(Spam) = 0.2

P(Free|Spam) = 0.6

P(Free) = 0.16

Then P(Spam|Free) = (0.6 × 0.2) / 0.16 = 0.75

```python
P_A = 0.2
P_B_given_A = 0.6
P_B = 0.16

P_A_given_B = (P_B_given_A * P_A) / P_B

print(P_A_given_B)
```

# Probability Rules

* Probability rules are basic principles used to calculate the likelihood of events.
* They help determine the probability of single events, combined events, and conditional events.
* Common rules include the addition rule, multiplication rule, and complement rule.

* Addition rule: P(A or B) = P(A) + P(B) - P(A and B)

* Multiplication rule (independent events): P(A and B) = P(A) × P(B)

* Complement rule: P(not A) = 1 - P(A)

### Example:

Rolling a dice

A = even number {2, 4, 6}, B = number greater than 4 {5, 6}, A and B = {6}

P(A or B) = 3/6 + 2/6 - 1/6 = 4/6

```python
P_even = 3/6
P_greater_4 = 2/6
P_both = 1/6

P_or = P_even + P_greater_4 - P_both
P_not_even = 1 - P_even

print("P(A or B) :", P_or)
print("P(not even) :", P_not_even)
```

# Random Variables

* A random variable is a variable that assigns a numerical value to each possible outcome of a random experiment.
* It helps us represent and analyze uncertain outcomes using numbers.

### Example:

Toss two coins and let x = number of heads

Possible values : 0, 1, 2

```python
import random

tosses = [random.choice(["H", "T"]) for _ in range(2)]

x = tosses.count("H")

print(tosses)
print("x =", x)
```

# Expected Value

* The expected value is the average value we expect from a random variable over many repeated trials.
* It is calculated by multiplying each possible outcome by its probability and adding all the results.

### Formula:

E(X) = Σ x × P(x)

### Example:

A game pays 0 with probability 0.5, 10 with probability 0.3 and 20 with probability 0.2.

E(X) = 0×0.5 + 10×0.3 + 20×0.2 = 7

```python
outcomes = [0, 10, 20]
probabilities = [0.5, 0.3, 0.2]

expected = sum(o * p for o, p in zip(outcomes, probabilities))

print("Expected value =", expected)
```
