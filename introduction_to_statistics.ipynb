# Descriptive Statistics
* It focuses on summarizing and organizing the data to make easily understand.
* common measures include:
* Mean
* Median
* Mode
* Minimum
* Maximum
* Range
* Variance
* Standard deviation
## Real time Example
* A company wants to get the
* highest salary
* lowest salary
* average salry
## example

```python
import numpy as np

salary = np.array([20000, 25000, 30000, 35000, 40000])

print("Mean:", np.mean(salary))
print("Median:", np.median(salary))
print("Minimum:", np.min(salary))
print("Maximum:", np.max(salary))
print("Range:", np.max(salary) - np.min(salary))
print("Variance:", np.var(salary))
print("Standard Deviation:", np.std(salary))
```

# Sample
* It is a smaller group selected in population
* Instead of collecting information from every person in a population we select some persons and study them
* It saves time and resources
## real time example
* instaed of collecting review from all the employees in a company
* we can take a sample review over a few employees


# Census
* Collecting Information from every employee
* There is no sampling involved because the complete population is studied.
* A census can provide highly complete information, but it can require significant time and resources.
## Real time example
* if a company has 100 employees and wants to know their average salary.
* If the company collects the salary of all 100 employees, it is a census.


# Sampling
* It  is the process of selecting a smaller group from a larger population.
* It is useful because studying the complete population may be expensive, slow, or impossible.
## real time example
An online shopping company has 1 million customers.
It wants to understand customer satisfaction.
Instead of contacting all 1 million customers, it randomly selects 10,000 customers.
This selection process is called sampling.
### example

```python
import pandas as pd

data = pd.DataFrame({
    "Age":[21,22,23,24,25,26,27,28]
})

sample = data.sample(5)

print(sample)
```

# Features
* A feature is an input variable used by a machine learning model to make predictions.
* it is a information given to an ML model to help it learn patterns.
## real time example
* fof example prdicting the house based on
* area of the house
* location
* number of bedrooms
### example


```python
import pandas as pd
data = {
    "Area": [1000, 1500, 2000],
    "Bedrooms": [2, 3, 4],
    "Age": [10, 5, 2],
    "Price": [50, 75, 100]
}
df = pd.DataFrame(data)
features = df[["Area", "Bedrooms", "Age"]]
print(features)
```
