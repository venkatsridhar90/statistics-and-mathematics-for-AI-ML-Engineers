#  String Operations



# 1.lower()


## Step 1 – Understand the Concept

The `lower()` method converts all uppercase letters in a string into lowercase letters.

For example:

"HELLO WORLD" → "hello world"

It is useful when we want to make text values consistent.

For example, the following values have the same meaning but different letter cases:

"Chennai"
"CHENNAI"
"chennai"

Converting them to lowercase makes comparison and text processing easier.

## Step 2 – Demonstrate the Concept

### Simple Example

"HELLO" becomes "hello" using `lower()`.

### Real-World Example

A website can convert user-entered email addresses into lowercase before processing them.

### Business Example

A company can convert customer city names into lowercase to avoid treating "Chennai" and "CHENNAI" as different values.

### AI/ML Use Case

Text preprocessing can convert words into lowercase before creating features for a machine learning model.

### Data Analytics Use Case

Analysts can standardize text categories before grouping and counting them.


## Step 3: Implement the Concept

```python
text = "HELLO PYTHON"

lower_text = text.lower()

print(lower_text)
```

## Code Explanation

- `lower()` converts uppercase letters into lowercase letters.
- The original string is not changed; a new lowercase string is returned.
- The output will be `hello python`.
- In AI/ML, it is commonly used during text preprocessing.
- In Data Analytics, it helps standardize text values before analysis.



# 2.upper()


## Step 1 – Understand the Concept

The `upper()` method converts all lowercase letters in a string into uppercase letters.

For example:

"hello world" → "HELLO WORLD"

It is useful when text needs to be displayed or standardized in uppercase format.

## Step 2 – Demonstrate the Concept

### Simple Example

"python" becomes "PYTHON" using `upper()`.

### Real-World Example

A system can convert country codes such as "in" into "IN".

### Business Example

A company can standardize product codes or department codes using uppercase letters.

### AI/ML Use Case

It can be used during text preprocessing when uppercase representation is required for a specific task.

### Data Analytics Use Case

It can standardize categorical text values before comparing or grouping them.


## Step 3: Implement the Concept

```python
text = "hello python"

upper_text = text.upper()

print(upper_text)
```

## Code Explanation

- `upper()` converts lowercase letters into uppercase letters.
- The original string remains unchanged.
- The output will be `HELLO PYTHON`.
- In AI/ML, it can be used as part of text preprocessing.
- In Data Analytics, it helps maintain consistent text formatting.



# 3.strip()


## Step 1 – Understand the Concept

The `strip()` method removes unwanted spaces or specified characters from the beginning and end of a string.

For example:

"  Python  " → "Python"

It does not remove spaces between words.

For example:

"Python Programming"

remains unchanged.

`strip()` is useful for cleaning text data before analysis.

## Step 2 – Demonstrate the Concept

### Simple Example

"  Hello  " becomes "Hello" after using `strip()`.

### Real-World Example

A user may accidentally enter spaces before or after their name in a registration form.

### Business Example

Customer data imported from different systems may contain unnecessary spaces around names or product names.

### AI/ML Use Case

Text cleaning can remove unwanted spaces before further preprocessing.

### Data Analytics Use Case

It helps clean categorical columns so that values with extra spaces are not treated as different categories.

## Step 3: Implement the Concept

```python
text = "   Python Programming   "

clean_text = text.strip()

print(clean_text)
```

## Code Explanation

- `strip()` removes leading and trailing spaces.
- It does not remove spaces between words.
- The output will be `Python Programming`.
- In AI/ML, it is useful during text cleaning.
- In Data Analytics, it helps clean inconsistent values in text columns.



# 4. split()


## Step 1 – Understand the Concept

The `split()` method divides a string into smaller parts and returns them as a list.

By default, it splits the string wherever there is whitespace.

For example:

"Python Pandas NumPy"

becomes:

["Python", "Pandas", "NumPy"]

We can also specify a separator such as a comma or hyphen.

## Step 2 – Demonstrate the Concept

### Simple Example

"Python Pandas NumPy" can be split into individual words.

### Real-World Example

A system can split a full name into first name and last name.

### Business Example

A company can split a product category string such as "Laptop,Phone,Tablet" into individual products.

### AI/ML Use Case

Text can be split into individual words or tokens as part of text preprocessing.

### Data Analytics Use Case

Analysts can split combined text values into separate parts for further analysis.


## Step 3: Implement the Concept

```python
text = "Python Pandas NumPy"

words = text.split()

print(words)
```

## Code Explanation

- `split()` divides a string into smaller parts.
- When no separator is provided, whitespace is used.
- The output is a list containing each word.
- In AI/ML, splitting text is useful during tokenization and preprocessing.
- In Data Analytics, it helps separate combined text values into individual elements.



# 5.contains()

## Step 1 – Understand the Concept

The `contains()` operation checks whether a specific text or pattern exists inside a string.

In Pandas, `str.contains()` is commonly used to search text values in a Series or DataFrame column.

It returns `True` when the specified text is found and `False` when it is not found.

This is useful for searching and filtering text data.

## Step 2 – Demonstrate the Concept

### Simple Example

We can check whether the word "Python" exists in a text value.

### Real-World Example

A search system can check whether a product name contains the word "Laptop".

### Business Example

A company can filter customer feedback containing words such as "delivery" or "payment".

### AI/ML Use Case

It can help identify text records containing specific keywords before further text processing.

### Data Analytics Use Case

Analysts can filter rows based on keywords found in customer comments or product descriptions.


## Step 3: Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Product": [
        "Python Course",
        "Data Science Course",
        "Python Programming",
        "Web Development"
    ]
})

result = df[df["Product"].str.contains("Python")]

print(result)
```

## Code Explanation

- `.str` provides string operations for a Pandas column.
- `.str.contains("Python")` checks whether "Python" exists in each value.
- It returns `True` for matching rows and `False` for non-matching rows.
- `df[...]` filters the DataFrame using those Boolean results.
- In AI/ML, it can help identify records containing specific keywords.
- In Data Analytics, it is useful for filtering and searching text data.



# 6.replace()

## Step 1 – Understand the Concept


The `replace()` method replaces one value or text with another value or text.

For example:

"Hello Python" → "Hello Pandas"

Here, "Python" is replaced with "Pandas".

It is useful for correcting, cleaning, and standardizing text data.

## Step 2 – Demonstrate the Concept

### Simple Example

"Python Course" can be changed to "Pandas Course" by replacing "Python" with "Pandas".

### Real-World Example

A customer database may contain an incorrect spelling that needs to be corrected.

### Business Example

A company may replace an old product name with a new product name across its dataset.

### AI/ML Use Case

Text preprocessing can replace unwanted words, symbols, or incorrect values before model training.

### Data Analytics Use Case

Analysts can replace inconsistent category names with standardized values.


## Step 3: Implement the Concept

```python
text = "I am learning Python"

updated_text = text.replace("Python", "Pandas")

print(updated_text)
```

## Code Explanation

- `replace()` searches for a specified value and replaces it with another value.
- `"Python"` is replaced with `"Pandas"`.
- The output will be `I am learning Pandas`.
- In AI/ML, it can be used during text cleaning and preprocessing.
- In Data Analytics, it helps correct and standardize text values.



# 7. Regular Expressions (Regex)

## Step 1 – Understand the Concept

Regular Expressions, or Regex, are patterns used to search, match, and extract specific text from strings.

Python provides the `re` module for working with regular expressions.

Regex can be used to find patterns such as:

- Email addresses
- Phone numbers
- Numbers
- Special characters
- Specific words or patterns

For example, the pattern `\d+` can be used to find one or more digits.

Regex is useful when simple string methods are not enough for complex text searching or extraction.

## Step 2 – Demonstrate the Concept

### Simple Example

From the text `"Order number is 12345"`, Regex can find the number `12345`.

### Real-World Example

A website can use Regex to identify email addresses entered into a registration form.

### Business Example

A company can extract phone numbers or order IDs from customer messages.

### AI/ML Use Case

Regex can be used during text preprocessing to remove unwanted patterns such as URLs, special characters, or numbers.

### Data Analytics Use Case

Analysts can extract useful patterns such as order IDs, product codes, or phone numbers from text columns.


## Step 3: Implement the Concept

```python
import re

text = "Order number is 12345"

pattern = r"\d+"

result = re.findall(pattern, text)

print(result)
```

## Code Explanation

- `import re` imports Python's Regular Expression module.
- `r"\d+"` is a Regex pattern.
- `\d` matches a digit.
- `+` means one or more occurrences.
- `re.findall()` finds all matching patterns in the text.
- The output will be `['12345']`.
- In AI/ML, Regex is useful for text cleaning and extracting patterns.
- In Data Analytics, it can extract structured information from unstructured text.
