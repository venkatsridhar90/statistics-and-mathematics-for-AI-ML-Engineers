# Statistics in Machine Learning

## 1. Data Cleaning
*  Data cleaning means finding and correcting incorrect, duplicate, inconsistent, or unwanted data.

### Real-time example:
* A customer dataset contains ages such as 25, 30, -5, and 200. Statistical checks can identify suspicious values.

### How statistics helps:
* Mean, median, minimum, maximum, and standard deviation help identify unusual or incorrect data values.

```python
import pandas as pd

df = pd.DataFrame({'Age': [25, 30, 28, -5, 200]})
print(df.describe())
```

## 2. Missing Value Handling

* Missing value handling means identifying empty values and deciding whether to remove or replace them.

### Real-time example:
* A customer dataset has missing income values, which can be replaced with the median income.

### How statistics helps:
* Mean, median, and mode provide statistically reasonable values for replacing missing observations.

```python
df = pd.DataFrame({'Income': [25000, 30000, None, 45000, 50000]})
df['Income'] = df['Income'].fillna(df['Income'].median())
print(df)
```

## 3. Outlier Detection
* An outlier is a value that is unusually far from most other observations.

### Real-time example:
* Most employee salaries are ₹30,000–₹80,000, but one salary is ₹5,00,000.

### How statistics helps:
* IQR, z-score, mean, and standard deviation help identify unusual observations.

```python
df = pd.DataFrame({'Salary': [30000, 35000, 40000, 42000, 45000, 50000, 500000]})
Q1 = df['Salary'].quantile(0.25)
Q3 = df['Salary'].quantile(0.75)
IQR = Q3 - Q1
outliers = df[(df['Salary'] < Q1 - 1.5*IQR) | (df['Salary'] > Q3 + 1.5*IQR)]
print(outliers)
```

## 4. Feature Engineering

*  Feature engineering means creating new useful variables from existing data to improve a model.

### Real-time example:
* From quantity and price, a company can create Total_Spend.

### How statistics helps:
* Statistics helps identify useful relationships and transformations between variables.

```python
df = pd.DataFrame({'Quantity': [2, 5, 3], 'Price': [100, 200, 150]})
df['Total_Spend'] = df['Quantity'] * df['Price']
print(df)
```

## 5. Feature Scaling

* Feature scaling puts numerical variables on a similar scale.

### Real-time example:
* Age and Salary have very different ranges, so scaling makes them comparable.

### How statistics helps:
* Standardization uses the mean and standard deviation.

```python
from sklearn.preprocessing import StandardScaler

X = [[25, 30000], [35, 60000], [45, 100000]]
X_scaled = StandardScaler().fit_transform(X)
print(X_scaled)
```

## 6. Data Normalization

* Normalization transforms values into a common range, usually 0 to 1.

### Real-time example:
* Ratings and purchase amounts can be converted to comparable ranges.

### How statistics helps:
* Minimum and maximum values are used for the transformation.

```python
from sklearn.preprocessing import MinMaxScaler

X = [[10, 100], [20, 500], [30, 1000]]
X_normalized = MinMaxScaler().fit_transform(X)
print(X_normalized)
```

## 7. Model Evaluation

*  Model evaluation measures how well a machine-learning model performs on unseen data.

### Real-time example:
* A bank checks how accurately its loan-default model predicts customers who may default.

### How statistics helps:
* Accuracy, precision, recall, F1-score, MAE, RMSE, and R² quantify performance.

```python
from sklearn.metrics import accuracy_score

y_test = [0, 1, 1, 0, 1]
y_pred = [0, 1, 0, 0, 1]
print("Accuracy:", accuracy_score(y_test, y_pred))
```

## 8. Linear Regression

*  Linear regression predicts a numerical value using relationships between input variables and a target.

### Real-time example:
* A company predicts sales from advertising expenditure.

### How statistics helps:
* Regression uses correlation, coefficients, variance, residuals, and hypothesis testing.

```python
from sklearn.linear_model import LinearRegression

X = [[10], [20], [30], [40], [50]]
y = [100, 150, 200, 250, 300]
model = LinearRegression().fit(X, y)
print("Prediction:", model.predict([[35]])[0])
```

## 9. Logistic Regression

*  Logistic regression predicts the probability of a categorical outcome such as Yes/No or 0/1.

### Real-time example:
* A bank predicts whether a customer will default on a loan.

### How statistics helps:
* It uses probability and statistical relationships to estimate class probabilities.

```python
from sklearn.linear_model import LogisticRegression

X = [[20], [25], [30], [35], [40], [45]]
y = [0, 0, 0, 1, 1, 1]
model = LogisticRegression().fit(X, y)
print("Predicted class:", model.predict([[32]])[0])
```

## 10. Naive Bayes

*  Naive Bayes is a classification algorithm based on Bayes' theorem and conditional probability.

### Real-time example:
* An email system predicts whether an email is spam or not spam.

### How statistics helps:
* Conditional probability is used to calculate the probability of each class.

```python
from sklearn.naive_bayes import GaussianNB

X = [[1], [2], [3], [8], [9], [10]]
y = [0, 0, 0, 1, 1, 1]
model = GaussianNB().fit(X, y)
print("Predicted class:", model.predict([[7]])[0])
```

## 11. Clustering

* Clustering groups similar observations together without predefined labels.

### Real-time example:
* An e-commerce company groups customers into budget, regular, and premium buyers.

### How statistics helps:
* Statistics helps measure similarity, distance, variation, and patterns within groups.

```python
from sklearn.cluster import KMeans

X = [[10, 20], [11, 21], [12, 22], [50, 60], [51, 61], [52, 62]]
model = KMeans(n_clusters=2, random_state=42, n_init=10)
print("Cluster labels:", model.fit_predict(X))
```

## 12. Principal Component Analysis (PCA)

* PCA reduces many variables into fewer important components while retaining as much information as possible.

### Real-time example:
* A company reduces 20 customer features to 2 or 3 components for easier analysis.

### How statistics helps:
* PCA uses variance and covariance to find directions containing the most information.

```python
from sklearn.decomposition import PCA

X = [[10, 20, 30], [12, 22, 32], [20, 40, 60], [25, 45, 65], [30, 50, 70]]
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)
print(X_pca)
print("Explained variance:", pca.explained_variance_ratio_)
```

## 13. Recommendation Systems

*  A recommendation system predicts products, movies, or services a user may like using behavior or similarity.

### Real-time example:
* Netflix recommends movies based on watched movies and similar users.

### How statistics helps:
* Statistics measures similarity, preferences, patterns, probabilities, and relationships.

```python
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd

user_item_matrix = pd.DataFrame({
    'Movie_A': [5, 4, 1],
    'Movie_B': [4, 5, 1],
    'Movie_C': [1, 1, 5]
}, index=['User_1', 'User_2', 'User_3'])

print(cosine_similarity(user_item_matrix))
```
