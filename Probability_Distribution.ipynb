# Probability Distribution
* A probability distribution describes how the probabilities of different possible values or outcomes are distributed.
* it tells us what outcomes are possible and how likely each outcome is.
### Why it is important
* Understands the data
* Make predictions
* Understand distribution
* Handle uncertainity
* Detects unusual values


# Bernoulli Distribution
* it is a probability distribution for an experiment that has only two possible outcomes
* Success = 1
* Failure = 0
* The experiment is performed only once.
### real time example
* A customer buying a product
* Buys a product --> 1
* does not buys a product --> 0
### formula



* X = outcome
* p = probability of success
* 1-p = probability of failure
* x = 0 or 1
### Example

If the probability that a customer buys a product is 0.7:

$$ P(X=1)=0.7 $$
$$ P(X=0)=1−0.7=0.3 $$ 

```python
import numpy as np

purchase = np.random.binomial(n=1, p=0.7)

print("Purchase:", purchase)
```

# Binomial Distribution
* A Binomial Distribution describes the number of successes obtained from a fixed number of independent trials,
* where each trial has only two possible outcomes.
* Unlike Bernoulli Distribution, Binomial Distribution involves multiple trials.
### real time example
* A company contacts 100 customers.
* Purchases the product
* Does not purchase the product
* The number of customers who purchase the product can be modeled using Binomial Distribution.
### Formula



n = number of trials

k = number of successes

p = probability of success

1-p = probability of failure

### example


```python
import numpy as np

data = np.random.binomial(10, 0.5)

print("Purchases :",data)
```

# Poisson Distribution
* A Poisson Distribution is used to model the number of times an event occurs within a fixed interval of time or space.
* The event should occur independently and at an approximately constant average rate.
### real time example
* A restaurant receives an average of 10 customers per hour.
* Poisson Distribution can be used to estimate the probability of receiving:
* 5 customers
* 10 customers
* 15 customers in one hour.
### Formula





```python
import numpy as np

calls = np.random.poisson(lam=5)
print("Calls received:", calls)
```

# Uniform Distribution
* A Uniform Distribution is a probability distribution where all values within a specified range have an equal probability.
* There are two types:
* Discrete Uniform Distribution
* Continuous Uniform Distribution
### real time example
Suppose an online shopping website provides a randomly generated discount between 0% and 10%.

Any discount between 0% and 10% can be generated with equal likelihood under a continuous uniform model.

### formula


* Mainly it is used for random number generating
* games
* scheduling
* testing syatems....


```python
import numpy as np

discount = np.random.uniform(10, 30)
print("Discount:", round(discount,2), "%")
```

# Normal Distribution
* A Normal Distribution is a continuous probability distribution with a bell-shaped curve.

* It is symmetric around the mean.

* In a perfectly normal distribution:

* Mean=Median=Mode

### Formula



### real time example
* Suppose the heights of a large group of people approximately follow a normal distribution.
* Most people have heights near the average, while very short and very tall people are less common.



```python
import numpy as np

purchase_amount = np.random.normal(loc=100, scale=20)
print("Purchase Amount:", round(purchase_amount,2))
```

# Exponential Distribution
* An Exponential Distribution is a continuous probability distribution commonly used to model the waiting time until an event occurs.
* It is closely related to the Poisson Distribution.
 Simple idea

* Poisson Distribution → Number of events

* Exponential Distribution → Waiting time between events
### real time example
* Suppose customers arrive at a restaurant randomly.
* The Exponential Distribution can be used to model the time until the next customer arrives.

### formula






```python
import numpy as np

waiting_time = np.random.exponential(scale=5)
print("Waiting Time:", round(waiting_time,2), "minutes")
```
