# sampling

What is sampling?

Sampling is the process of selecting a smaller group from a larger population for analysis.
It helps researchers study data efficiently without examining the entire population.
A properly selected sample can provide meaningful and reliable insights about the population.
It saves time, cost, and effort while conducting data analysis.

Example:
    
If a college has 2,000 students and we select 200 students for a survey, the 200 selected students are called the sample.
The 2,000 students represent the population, and selecting 200 students is called sampling.


```python
import random

students = list(range(1, 2001))

sample = random.sample(students, 200)
print(sample[:10]) # Display the first 10 selected students
```

[128, 897, 23, 9, 1924, 278, 964, 1532, 83, 562]

A sample of 200 students is selected from the population of 2,000 students


Random Sampling

Random sampling is a technique in which individuals are selected randomly from a population.
Each individual has an equal opportunity to be included in the sample.
It is useful for obtaining a fair and unbiased representation of the population.

Example:

The principal randomly selects 200 students from the list of 2,000 students.


```python
import random

students = list(range(1, 2001))

random_sample = random.sample(students, 200)
print(random_sample[:10])
```

[439, 454, 235, 795, 734, 967, 452, 656, 356, 190]

Every student has an equal probability of being selected.



Stratified Sampling

Stratified sampling is a method of dividing a population into different groups based on common characteristics.
These groups are called strata.
A sample is then selected from each stratum to ensure proper representation.
It helps provide a more accurate representation of the entire population.

Example:

The college has students from:

* First Year
* Second Year
* Third Year
* Fourth Year

The principal selects 50 students from each year, resulting in 200 students.

```python

first_year = list(range(1, 251))
second_year = list(range(251, 501))
third_year = list(range(501, 751))
fourth_year = list(range(751, 1001))

import random

sample = (
    random.sample(first_year, 50) +
    random.sample(second_year, 50) +
    random.sample(third_year, 50) +
    random.sample(fourth_year, 50)
)

print(sample[:10])
```

[55, 129, 30, 168, 87, 11, 48, 12, 92, 81]

Every academic year is represented equally in the sample.
    

Cluster Sampling

Cluster Sampling is a sampling method in which the population is divided into groups called clusters.
A few clusters are randomly selected from the total population.
All members of the selected clusters are included in the study.

Example:

The college has 40 classrooms.

The principal randomly selects 4 calssrooms and surveys every studnet in those clasrooms.

```python
import random

clusters = ["Class A", "Class B", "Class C", "Class D", "Class E","Class F",
            "Class G","Class H"]

selected_cluster = random.choice(clusters)
print(selected_cluster)
```

Class A

Instead of selecting individual students, entire classrooms are selected

Systematic Sampling

Systematic sampling is a probability sampling method.
The population is arranged in a specific order or list.
A random starting point is selected, followed by every kth member.
It provides a simple and organized way to select a sample.

Example:

The principa; selects every 20th student from the student register.

```python
students = list(range(1, 2001))

systematic_sample = students[::20]

print(systematic_sample)[:20])
```

[1, 11, 21, 31, 41, 51, 61, 71, 81, 91,
 101, 111, 121, 131, 141, 151, 161, 171, 181, 191]

Students are selected at regular intervals.


Sampling Bias

Sampling bias occurs when a sample does not fairly represent the population.
Some members may have a higher or lower chance of being selected.
This can lead to inaccurate or misleading study results.
Proper sampling methods can help reduce sampling bias.

Example

The principal surveys only third-year students about the canteen.

The opinions of first-,second-,and final-year students are ignored.


```python
third_year_students = list(range(501, 751))
sample = third_year_students[:20]
```

[501, 502, 503, 504, 505, 506, 507, 508, 509, 510,
 511, 512, 513, 514, 515, 516, 517, 518, 519, 520]

The sample does not represent the entire student population.

Sampling Error

Sampling error is the difference between the sample result and the true population result.
It occurs because only a part of the population is studied.
The result may differ from what would be obtained by studying the entire population.
Larger and properly selected samples can help reduce sampling error.

Common Example

Suppose:
* Population average satisfaction = 8.5
* Sample average satisfaction = 8.1

The difference (0.4) is the sampling error.


```python
population_mean = 8.5
sample_mean = 8.1

sampling_error = population_mean - sample_mean

print(sampling_error)
```

A small sampling error indicates that the sample closely represents the population.

Sample Size

Sample size is the number of individuals selected from a population for a study.
It represents the portion of the population included in the research.
A suitable sample size helps produce reliable and accurate results.

Example:
* Population =2,000 Students
* Sample Size =200  Students

```python
students = list(range(1, 2001))

sample = students[:200]

print("Sample Size:", len(sample))
```

The survey is conducted using 200 students out of the total 2,000 students.
