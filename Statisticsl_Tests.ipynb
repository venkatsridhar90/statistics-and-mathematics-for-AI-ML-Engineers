# Statistical tests



# 1. T-Test
* A t-test checks whether the means of two groups are significantly different
*  It is commonly used when the population standard deviation is unknown.
* Data are independent, approximately normally distributed, and measured on a continuous scale.
*  For a standard independent t-test, equal variance is assumed unless Welch’s t-test is used.
#### When to use:
Use it to compare the means of two groups, such as average sales before and after a change or between two teams.
#### When not to use:
* Do not use it for more than two independent groups or strongly non-normal data with severe outliers;consider ANOVA or a non-parametric test.
#### Business Example: 
* A company compares average monthly sales of employees using a new sales script versus the old script.

### Interpretation:
* If p < 0.05, reject the null hypothesis and conclude the group means are significantly different. If p ≥
0.05, there is not enough evidence of a difference.


```python
from scipy.stats import ttest_ind

group_A = [75, 80, 85, 78, 82]
group_B = [70, 72, 68, 75, 71]

t_stat, p = ttest_ind(group_A, group_B, equal_var=False)

print("T-statistic:", t_stat)
print("P-value:", p)
```

# 2. Z-Test
* A z-test tests whether a sample mean or proportion differs significantly from a known or hypothesized
value.
* It is mainly used with known population variance or large samples.
* Observations should be independent and the sampling distribution should be approximately normal.
* For a mean z-test, the population standard deviation is typically known.
#### When to use:
* Use it for large-sample mean or proportion comparisons when the population standard deviation is
known or a z approximation is appropriate.
#### When not to use:
* Avoid it for small samples when population standard deviation is unknown; a t-test is generally more
appropriate.
#### Business Example: 
* A bank checks whether the average transaction amount differs from the historical average of
n5,000 using a large sample and known population variation.

#### Interpretation: 
* If p < 0.05, the sample provides evidence that the mean differs from the hypothesized value.
Otherwise, the difference is not statistically significant.


```python
from statsmodels.stats.weightstats import ztest

sample = [5100, 5200, 4950, 5050, 5150]

z_stat, p = ztest(sample, value=5000)

print("Z-statistic:", z_stat)
print("P-value:", p)
```

# 3. Chi-Square Test
*  A chi-square test analyzes categorical count data.
* It is commonly used to test independence between two
categorical variables or goodness of fit.
*  Observations should be independent and data should be frequencies/counts.
*   Expected cell counts should generally be large enough for the chi-square approximation to be reliable.
#### When to use:
 * Use it to determine whether two categorical variables are associated, such as product preference and
customer age group.
#### When not to use:
* Do not use it for continuous measurements or very small expected frequencies
*  Fisher’s exact test
may be better for small 2×2 tables.
#### Business Example:
*  A retailer tests whether customer purchase preference is associated with customer age group.

#### Interpretation: 
* If p < 0.05, reject independence and conclude the categorical variables are significantly associated. \
* If p
≥ 0.05, there is insufficient evidence of association.


```python
from scipy.stats import chi2_contingency

table = [
    [30, 20],
    [10, 40]
]

chi2, p, dof, expected = chi2_contingency(table)

print("Chi-Square Statistic:", chi2)
print("P-value:", p)
print("Degrees of Freedom:", dof)
print("Expected Frequencies:")
print(expected)
```

# 4. ANOVA
* ANOVA (Analysis of Variance) tests whether the means of three or more groups are significantly different.
* It compares between-group variation with within-group variation.
*  Observations should be independent, groups should be approximately normally distributed, and
variances should be reasonably similar.
* The outcome should be continuous.
### When to use: 
* Use it when comparing the average outcome across three or more independent groups, such as sales
across several regions.
### When not to use:
* Do not use ordinary one-way ANOVA when assumptions are badly violated or observations are
dependent
* consider Kruskal-Wallis or repeated-measures methods.
### Business Example:
* A company compares average customer satisfaction scores across four branches.

### Interpretation: 
* If p < 0.05, at least one group mean differs from another.
* ANOVA does not identify which groups differ;
use a post-hoc test such as Tukey’s test.


```python
from scipy.stats import f_oneway

branch_A = [80, 85, 82, 88, 90]
branch_B = [75, 78, 80, 76, 79]
branch_C = [90, 92, 88, 94, 91]
branch_D = [70, 72, 75, 68, 74]

f_stat, p = f_oneway(branch_A, branch_B, branch_C, branch_D)

print("F-statistic:", f_stat)
print("P-value:", p)
```

# 5. Mann-Whitney U Test
* The Mann-Whitney U test is a non-parametric test for comparing two independent groups.
*  It evaluates
whether one group tends to have higher values than the other.
 * The groups are independent and observations are ordinal or continuous.
 *  It does not require normality,
but similar distribution shapes are useful when interpreting it as a difference in location.
#### When to use:
* Use it when comparing two independent groups and the data are ordinal, skewed, or contain
problematic outliers.
#### When not to use:
* Do not use it for paired/dependent observations; use the Wilcoxon signed-rank test for paired data.
#### Business Example:
* A company compares customer waiting-time ratings between two service centers when the
ratings are strongly skewed.

#### Interpretation: 
* If p < 0.05, the groups show a statistically significant difference in their distributions/ranks. 
* If p ≥ 0.05,
evidence of a difference is insufficient.


```python
from scipy.stats import mannwhitneyu

group_A = [75, 80, 85, 78, 82]
group_B = [70, 72, 68, 75, 71]

u, p = mannwhitneyu(group_A, group_B, alternative='two-sided')

print("U-statistic:", u)
print("P-value:", p)
```

# 6. Shapiro-Wilk Test
* The Shapiro-Wilk test checks whether a sample is consistent with a normal distribution.
*  It is commonly
used as a normality check before parametric tests.
*  Observations should be independent and numeric.
*  The test is sensitive to sample size, so the p-value
should be considered along with plots such as a Q-Q plot.
### When to use:
*  Use it to assess normality when deciding whether a t-test or ANOVA’s normality assumption is
reasonable.
### When not to use:
*  Do not rely on it alone for very large samples, where tiny deviations may become significant;
combine it with visual diagnostics and subject-matter knowledge.
### Business Example:
* Before running a t-test on employee productivity scores, an analyst checks whether the scores
are approximately normally distributed.
### Interpretation:
* If p < 0.05, reject the normality assumption and investigate non-normality.
*  If p ≥ 0.05, there is
insufficient evidence to say the data are non-normal

```python
from scipy.stats import shapiro

data = [12, 15, 14, 13, 16, 15, 14, 17, 13, 15]

w, p = shapiro(data)

print("W-statistic:", w)
print("P-value:", p)
```
